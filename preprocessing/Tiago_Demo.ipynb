### Demonstration of Resizing and Cropping



```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from resizing_64 import resizing_to_64
from cropp_face import cropping_face
from load_datasets import CustomDatasetLoaderwithoutResizing
```

```python
label_map = {
    1: "surprise",
    2: "fear",
    3: "disgust",
    4: "happy",
    5: "sad",
    6: "angry"
}

loader = CustomDatasetLoaderwithoutResizing()
images,labels = loader.load_dataset() #I have not used the max argument due to the small size of the test


```

## Resizing the images

```python
for i in range(5):
    image = images[i]
    label = labels[i]

    w, h = image.size
    

    plt.tight_layout()
    ax = plt.subplot(1, 5, i + 1)
    ax.imshow(image)
    ax.set_title(f"size: {w}x{h} \n label: {label_map[label]}")
    ax.axis("on")


plt.show()
```
